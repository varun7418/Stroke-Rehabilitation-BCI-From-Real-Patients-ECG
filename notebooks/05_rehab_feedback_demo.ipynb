# 05 · Rehab feedback demo

## Like you're five
The computer guesses which hand you imagined, then a cartoon hand **squeezes shut** to say "I understood you!" It's a pretend game built from an old recording. It is **not** a medicine or a real therapy machine.

```
# --- Setup (works on your laptop AND on Google Colab) ---
import os, sys
if "google.colab" in sys.modules and not os.path.exists("src"):
    # Colab: get the project first, e.g.  !git clone <your-repo-url> && %cd stroke-rehab-bci/notebooks
    pass
sys.path.append("..")          # so we can "import src"
# !pip install -q moabb mne scikit-learn pandas matplotlib   # uncomment on Colab
import numpy as np, pandas as pd, matplotlib.pyplot as plt
```

```
sys.path.append("../app")
from src.preprocessing import load_liu2024
from src.models import get_models
from rehab_demo import render_feedback
X, y, meta = load_liu2024(subjects=list(range(1, 11)))
subj = meta["subject"].to_numpy(); held_out = subj[-1]
model = get_models()["CSP + LogReg"]
model.fit(X[subj != held_out], y[subj != held_out])     # never saw this patient
trial = X[subj == held_out][0:1]; true = y[subj == held_out][0]
pred = model.predict(trial)[0]; conf = model.predict_proba(trial).max()
print("held-out patient:", held_out, "| predicted:", pred, "| really:", true, "| confidence:", round(conf, 2))
render_feedback(pred, conf, true, save_path="../figures/rehab_demo.png")
plt.show()
```

A real rehab BCI could connect decoded intent to visual feedback, a robot, or electrical stimulation. This project only **simulates** the visual part, offline.
