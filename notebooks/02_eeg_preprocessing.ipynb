# 02 · Preprocessing: epochs and the 8–30 Hz filter

## Like you're five
- **Epoch** = one little slice of recording for one try. "Trial 001 = 4 seconds → LEFT."
- **Filter** = a sieve. The brain makes wiggles that are slow and wiggles that are fast. Movement thinking shows up in the **medium-speed** wiggles (**mu** 8–13 Hz and **beta** 13–30 Hz). We keep those and throw the rest away.

The filter does **not** magically find "movement thoughts." It only removes stuff outside the range we chose.

```
from src.preprocessing import load_liu2024
from scipy.signal import welch
X_wide, y, meta = load_liu2024(subjects=[1], fmin=1, fmax=40)   # almost unfiltered
X_filt, _, _   = load_liu2024(subjects=[1], fmin=8, fmax=30)    # our band
print(X_wide.shape, X_filt.shape)
```

```
f, p1 = welch(X_wide[:, 0, :].mean(0), fs=250, nperseg=250)
f, p2 = welch(X_filt[:, 0, :].mean(0), fs=250, nperseg=250)
plt.semilogy(f, p1, label="1-40 Hz"); plt.semilogy(f, p2, label="8-30 Hz (ours)")
plt.axvspan(8, 13, alpha=.15, label="mu"); plt.axvspan(13, 30, alpha=.08, color="orange", label="beta")
plt.xlim(0, 45); plt.xlabel("Frequency (Hz)"); plt.ylabel("Power"); plt.legend()
plt.title("What the filter keeps")
```

**Why 250 Hz is enough:** you need to sample at least twice the highest frequency you care about (Nyquist). 30 Hz × 2 = 60 Hz, so 250 Hz is plenty.
